```python
# === SITE SPECIFIC INPUTS ===
airflow_ls = 500        # Real-time or TAB measured L/s
sat = 27.41           # supply air temperature from peak
rat = 24.02     # return air temperature from peak
hours_leaking = 12
boiler_efficiency = 0.85    # 85% efficiency ( we need to find the real value on site)
cost_per_kwh = 0.05

# === THERMODYNAMIC CONSTANTS (Metric)
air_density = 1.204      # kg/m3 (the common value for cities in the UK)
cp_air = 1.006           # kJ/kg.K (Specific heat)

# === CALCULATIONS ===
# Power (kW) = Mass Flow (kg/s) * Cp * Delta T
# Mass Flow = (L/s / 1000) * density
mass_flow = (airflow_ls / 1000) * air_density
kw_thermal = mass_flow * cp_air * (sat - rat)

# Total Energy Lost (Accounting for Boiler Efficiency)
kw_fuel_input = kw_thermal / boiler_efficiency
kwh_day = kw_fuel_input * hours_leaking

# Financials
daily_loss = kwh_day * cost_per_kwh
annual_projection = daily_loss * 365

print("--- Input Parameters ---")
print(f"Airflow: {airflow_ls} L/s")
print(f"Supply Air Temp (SAT): {sat}°C")
print(f"Return Air Temp (RAT): {rat}°C")
print(f"Hours Leaking Daily: {hours_leaking} hours")
print(f"Boiler Efficiency: {boiler_efficiency*100:.0f}%")
print(f"Cost per kWh: £{cost_per_kwh:.2f}")
print("------------------------")
print(f"Verified Thermal Power Loss: {kw_thermal:.2f} kW")
print(f"Equivalent Fuel Input (at {boiler_efficiency*100:.0f}% efficiency): {kw_fuel_input:.2f} kW")
print(f"Daily Energy Loss (operating {hours_leaking} hours): {kwh_day:.2f} kWh")
print(f"Daily Cost of Leak: £{daily_loss:.2f}")
print(f"Annual Cost of Leak: £{annual_projection:.2f}")
```
