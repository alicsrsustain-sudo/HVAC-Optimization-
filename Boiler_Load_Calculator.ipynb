```python
"""
Boiler Load & Cost Calculator
Compare 'Current' and 'Proposed' scenarios to see the effect of changing variables.
"""

# ─────────────────────────────────────────────
#  INPUTS
# ─────────────────────────────────────────────

# Shared / fixed properties
yearly_hours_run     = 1600    # hours/year
flowrate             = 0.002   # m³/s
specific_heat        = 4.2     # kJ/kg/K
density              = 997     # kg/m³
gas_unit_price       = 0.03    # £ per kWh
boiler_efficiency    = 90      # %

# Current scenario
Current_return_temp   = 75      # °C
Current_flow_temp     = 80      # °C

# Proposed scenario  ← tweak these to explore savings
Proposed_return_temp    = 76      # °C
Proposed_flow_temp      = 80      # °C


# ─────────────────────────────────────────────
#  CALCULATIONS
# ─────────────────────────────────────────────

def calculate(return_temp, flow_temp,
              flowrate, specific_heat, density,
              yearly_hours_run, boiler_efficiency, gas_unit_price):
    """Return a dict of all calculated outputs for one scenario."""
    delta_t        = flow_temp - return_temp                        # °C
    heat_load      = flowrate * density * specific_heat * delta_t   # kW  (kJ/s)
    gas_load       = heat_load / (boiler_efficiency / 100)          # kW
    yearly_gas_kwh = gas_load * yearly_hours_run                    # kWh/year
    yearly_cost    = yearly_gas_kwh * gas_unit_price                # £/year

    return {
        "Delta T (°C)"         : delta_t,
        "Heat Load (kW)"       : heat_load,
        "Gas Load (kW)"        : gas_load,
        "Yearly Gas Load (kWh)": yearly_gas_kwh,
        "Yearly Cost (£)"      : yearly_cost,
    }

Current = calculate(Current_return_temp, Current_flow_temp,
                   flowrate, specific_heat, density,
                   yearly_hours_run, boiler_efficiency, gas_unit_price)

Proposed  = calculate(Proposed_return_temp,  Proposed_flow_temp,
                   flowrate, specific_heat, density,
                   yearly_hours_run, boiler_efficiency, gas_unit_price)


# ─────────────────────────────────────────────
#  OUTPUT
# ─────────────────────────────────────────────

col_w = 26   # column width for alignment

print("=" * 70)
print("  BOILER LOAD & COST CALCULATOR")
print("=" * 70)

# --- Inputs summary ---
print("\n  INPUTS")
print("-" * 70)
print(f"  {'Parameter':<{col_w}} {'Current':>12}   {'Proposed':>12}   {'Unit'}")
print(f"  {'-'*col_w} {'------':>12}   {'-----':>12}   {'----'}")
print(f"  {'Yearly Hours Run':<{col_w}} {yearly_hours_run:>12}   {yearly_hours_run:>12}   hours")
print(f"  {'Flowrate':<{col_w}} {flowrate:>12.4f}   {flowrate:>12.4f}   m³/s")
print(f"  {'Specific Heat Capacity':<{col_w}} {specific_heat:>12.1f}   {specific_heat:>12.1f}   kJ/kg/K")
print(f"  {'Density':<{col_w}} {density:>12}   {density:>12}   kg/m³")
print(f"  {'Return Temp':<{col_w}} {Current_return_temp:>12}   {Proposed_return_temp:>12}   °C")
print(f"  {'Flow Temp':<{col_w}} {Current_flow_temp:>12}   {Proposed_flow_temp:>12}   °C")
print(f"  {'Gas Unit Price':<{col_w}} {gas_unit_price:>12.4f}   {gas_unit_price:>12.4f}   £/kWh")
print(f"  {'Boiler Efficiency':<{col_w}} {boiler_efficiency:>12}   {boiler_efficiency:>12}   %")

# --- Results ---
print("\n  RESULTS")
print("-" * 70)
print(f"  {'Metric':<{col_w}} {'Current':>12}   {'Proposed':>12}   {'Unit'}")
print(f"  {'-'*col_w} {'------':>12}   {'-----':>12}   {'----'}")

units = {
    "Delta T (°C)"         : "°C",
    "Heat Load (kW)"       : "kW",
    "Gas Load (kW)"        : "kW",
    "Yearly Gas Load (kWh)": "kWh",
    "Yearly Cost (£)"      : "£",
}

for key in Current:
    b = Current[key]
    a = Proposed[key]
    u = units[key]
    print(f"  {key:<{col_w}} {b:>12.2f}   {a:>12.2f}   {u}")

# --- Savings summary ---
kwh_saving  = Current["Yearly Gas Load (kWh)"] - Proposed["Yearly Gas Load (kWh)"]
cost_saving = Current["Yearly Cost (£)"]       - Proposed["Yearly Cost (£)"]
pct_saving  = (cost_saving / Current["Yearly Cost (£)"]) * 100

print("\n  SAVINGS")
print("-" * 70)
print(f"  {'Energy Saving':<{col_w}} {kwh_saving:>12.2f}   kWh/year")
print(f"  {'Cost Saving':<{col_w}} {cost_saving:>12.2f}   £/year")
print(f"  {'Cost Saving (%)':<{col_w}} {pct_saving:>12.1f}   %")
print("=" * 70)
```
