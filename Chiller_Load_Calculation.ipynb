```python
"""
Chiller Load & Cost Calculator
Compare 'Current' and 'Proposed' scenarios to see the effect of changing variables.
"""

# ─────────────────────────────────────────────
#  INPUTS
# ─────────────────────────────────────────────

# Shared / fixed properties
yearly_hours_run     = 3650    # hours/year
flowrate             = 0.002   # m³/s
specific_heat        = 4.2     # kJ/kg/K
density              = 997     # kg/m³
electricity_price    = 0.20    # £ per kWh
chiller_cop          = 4       # Coefficient of Performance

# Current scenario
current_return_temp  = 12      # °C
current_flow_temp    = 8       # °C

# Proposed scenario  ← tweak these to explore savings
proposed_return_temp = 11      # °C
proposed_flow_temp   = 8       # °C


# ─────────────────────────────────────────────
#  CALCULATIONS
# ─────────────────────────────────────────────

def calculate(return_temp, flow_temp,
              flowrate, specific_heat, density,
              yearly_hours_run, chiller_cop, electricity_price):
    """Return a dict of all calculated outputs for one scenario."""
    delta_t            = return_temp - flow_temp                        # °C (warm return minus cold flow)
    cooling_load       = flowrate * density * specific_heat * delta_t   # kW
    electrical_load    = cooling_load / chiller_cop                     # kW
    yearly_elec_kwh    = electrical_load * yearly_hours_run             # kWh/year
    yearly_cost        = yearly_elec_kwh * electricity_price            # £/year

    return {
        "Delta T (°C)"               : delta_t,
        "Cooling Load (kW)"          : cooling_load,
        "Electrical Load (kW)"       : electrical_load,
        "Yearly Electrical Load (kWh)": yearly_elec_kwh,
        "Yearly Cost (£)"            : yearly_cost,
    }

current  = calculate(current_return_temp,  current_flow_temp,
                     flowrate, specific_heat, density,
                     yearly_hours_run, chiller_cop, electricity_price)

proposed = calculate(proposed_return_temp, proposed_flow_temp,
                     flowrate, specific_heat, density,
                     yearly_hours_run, chiller_cop, electricity_price)


# ─────────────────────────────────────────────
#  OUTPUT
# ─────────────────────────────────────────────

col_w = 30   # column width for alignment

print("=" * 74)
print("  CHILLER LOAD & COST CALCULATOR")
print("=" * 74)

# --- Inputs summary ---
print("\n  INPUTS")
print("-" * 74)
print(f"  {'Parameter':<{col_w}} {'Current':>12}   {'Proposed':>12}   {'Unit'}")
print(f"  {'-'*col_w} {'-------':>12}   {'--------':>12}   {'----'}")
print(f"  {'Yearly Hours Run':<{col_w}} {yearly_hours_run:>12}   {yearly_hours_run:>12}   hours")
print(f"  {'Flowrate':<{col_w}} {flowrate:>12.4f}   {flowrate:>12.4f}   m³/s")
print(f"  {'Specific Heat Capacity':<{col_w}} {specific_heat:>12.1f}   {specific_heat:>12.1f}   kJ/kg/K")
print(f"  {'Density':<{col_w}} {density:>12}   {density:>12}   kg/m³")
print(f"  {'Return Temp':<{col_w}} {current_return_temp:>12}   {proposed_return_temp:>12}   °C")
print(f"  {'Flow Temp':<{col_w}} {current_flow_temp:>12}   {proposed_flow_temp:>12}   °C")
print(f"  {'Electricity Unit Price':<{col_w}} {electricity_price:>12.4f}   {electricity_price:>12.4f}   £/kWh")
print(f"  {'Chiller COP':<{col_w}} {chiller_cop:>12}   {chiller_cop:>12}   -")

# --- Results ---
print("\n  RESULTS")
print("-" * 74)
print(f"  {'Metric':<{col_w}} {'Current':>12}   {'Proposed':>12}   {'Unit'}")
print(f"  {'-'*col_w} {'-------':>12}   {'--------':>12}   {'----'}")

units = {
    "Delta T (°C)"               : "°C",
    "Cooling Load (kW)"          : "kW",
    "Electrical Load (kW)"       : "kW",
    "Yearly Electrical Load (kWh)": "kWh",
    "Yearly Cost (£)"            : "£",
}

for key in current:
    c = current[key]
    p = proposed[key]
    u = units[key]
    print(f"  {key:<{col_w}} {c:>12.2f}   {p:>12.2f}   {u}")

# --- Savings summary ---
kwh_saving  = current["Yearly Electrical Load (kWh)"] - proposed["Yearly Electrical Load (kWh)"]
cost_saving = current["Yearly Cost (£)"]               - proposed["Yearly Cost (£)"]
pct_saving  = (cost_saving / current["Yearly Cost (£)"]) * 100

print("\n  SAVINGS")
print("-" * 74)
print(f"  {'Energy Saving':<{col_w}} {kwh_saving:>12.2f}   kWh/year")
print(f"  {'Cost Saving':<{col_w}} {cost_saving:>12.2f}   £/year")
print(f"  {'Cost Saving (%)':<{col_w}} {pct_saving:>12.1f}   %")
print("=" * 74)
```
