```python

# --- CUSTOMIZE YOUR VARIABLES HERE ---
INITIAL_POWER = 15.0        # Motor power at full speed (in kW)
FULL_SPEED = 1800           # Rated full speed (in RPM or %)
SUGGESTED_SPEED = 1440      # Your lower suggested speed (in RPM or %)
ELECTRICITY_RATE = 0.25     # Cost per kWh (e.g., £0.15)

def calculate_affinity_savings(
    initial_power_kw,
    initial_speed_rpm,
    new_speed_rpm,
    electricity_rate_per_kwh,
    hours_per_day=12,
    days_per_week=5,
    weeks_per_year=52
):
    # 1. Calculate the speed rastio
    speed_ratio = new_speed_rpm / initial_speed_rpm

    # 2. Apply the Affinity Law for Power: (N2/N1)^3
    new_power_kw = initial_power_kw * (speed_ratio ** 3)

    # 3. Calculate operating hours
    weekly_hours = hours_per_day * days_per_week
    annual_hours = weekly_hours * weeks_per_year

    # 4. Calculate energy consumption (kWh)
    initial_kwh_yearly = initial_power_kw * annual_hours
    new_kwh_yearly = new_power_kw * annual_hours

    # 5. Calculate savings
    energy_saved_kwh_yearly = initial_kwh_yearly - new_kwh_yearly
    money_saved_yearly = energy_saved_kwh_yearly * electricity_rate_per_kwh

    # Print the breakdown
    print("=" * 45)
    print("         AFFINITY LAW SAVINGS REPORT         ")
    print("=" * 45)
    print(f"Speed Reduction:       {initial_speed_rpm} RPM -> {new_speed_rpm} RPM ({(1 - speed_ratio)*100:.1f}% reduction)")
    print(f"Power Demand:          {initial_power_kw:.2f} kW -> {new_power_kw:.2f} kW")
    print(f"Operating Profile:     {hours_per_day} hrs/day, {days_per_week} days/week ({annual_hours} hrs/year)")
    print("-" * 45)
    print(f"Annual Energy Saved:   {energy_saved_kwh_yearly:,.2f} kWh")
    print(f"Annual Money Saved:    £{money_saved_yearly:,.2f}")
    print("=" * 45)

# Run the calculation
calculate_affinity_savings(
    initial_power_kw=INITIAL_POWER,
    initial_speed_rpm=FULL_SPEED,
    new_speed_rpm=SUGGESTED_SPEED,
    electricity_rate_per_kwh=ELECTRICITY_RATE,
    hours_per_day=12,
    days_per_week=5
)
```
