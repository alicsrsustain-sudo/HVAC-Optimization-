```python
# === SITE SPECIFIC INPUTS ===
airflow_ls = 500        # Real-time or TAB measured L/s
sat = 14.00             # Supply air temperature from PEAK (°C) - typically 12-16°C for chilled
rat = 24.02             # Return air temperature from PEAK (°C)
hours_leaking = 12
chiller_cop = 3.5       # Chiller COP (typical range: 2.5 - 5.0, find real value on site)
cost_per_kwh = 0.25     # Electricity cost (higher than gas - chillers are electrically driven)

# === THERMODYNAMIC CONSTANTS (Metric) ===
air_density = 1.204     # kg/m3 (common value for UK cities)
cp_air = 1.006          # kJ/kg.K (Specific heat of air)

# === CALCULATIONS ===
# Mass Flow = (L/s / 1000) * density
mass_flow = (airflow_ls / 1000) * air_density

# Thermal load (kW) = Mass Flow (kg/s) * Cp * Delta T
# Delta T = RAT - SAT (RAT is warmer, SAT is cold for chilled air)
delta_t = rat - sat
kw_thermal = mass_flow * cp_air * delta_t

# Chiller electrical draw (accounting for COP)
# COP = Thermal Load / Electrical Input  =>  Electrical Input = Thermal Load / COP
kw_electrical = kw_thermal / chiller_cop
kwh_day = kw_electrical * hours_leaking

# Financials
daily_loss = kwh_day * cost_per_kwh
annual_projection = daily_loss * 365

print("--- Input Parameters ---")
print(f"Airflow: {airflow_ls} L/s")
print(f"Supply Air Temp (SAT): {sat}°C")
print(f"Return Air Temp (RAT): {rat}°C")
print(f"Hours Leaking Daily: {hours_leaking} hours")
print(f"Chiller COP: {chiller_cop}")
print(f"Cost per kWh: £{cost_per_kwh:.2f}")
print("------------------------")
print(f"Delta T (RAT - SAT): {delta_t:.2f}°C")
print(f"Verified Thermal Load: {kw_thermal:.2f} kW")
print(f"Chiller Electrical Draw (at COP {chiller_cop}): {kw_electrical:.2f} kW")
print(f"Daily Energy Loss (operating {hours_leaking} hours): {kwh_day:.2f} kWh")
print(f"Daily Cost of Leak: £{daily_loss:.2f}")
print(f"Annual Cost of Leak: £{annual_projection:.2f}")
```
