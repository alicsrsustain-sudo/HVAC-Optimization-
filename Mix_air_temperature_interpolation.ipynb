```python
"""
Mixed Air Temperature Calculator
Mixed Air Temp = (Recirc Damper % × Return Temp) + (Fresh Air Damper % × Fresh Air Temp)
"""

# ── INPUTS ────────────────────────────────────────────────────────────────────
recirc_damper  = 0.5   # Recirculation damper position
fresh_damper   = 0.5   # Fresh air damper position
return_temp    = 19     # Return air temperature (°C)
fresh_air_temp = -1     # Fresh air (outdoor) temperature (°C) in here we can also import the yearly weather data

# ── VALIDATION ────────────────────────────────────────────────────────────────
if round(recirc_damper + fresh_damper, 6) != 1.0:
    raise ValueError(f"Damper positions must add up to 1.0 (got {recirc_damper + fresh_damper})")
  #In this part we are just double checking that the numbers typed in actually make sense. The number 6 tells the PC to round around 6 decimal places.

# ── CALCULATION ───────────────────────────────────────────────────────────────
mixed_air_temp = (recirc_damper * return_temp) + (fresh_damper * fresh_air_temp)

# ── OUTPUT ────────────────────────────────────────────────────────────────────
print("=" * 40) # This is used to beautify the result with those separating lines.
print("  MIXED AIR TEMPERATURE CALCULATOR")
print("=" * 40)
print(f"  Recirc Damper   : {recirc_damper * 100:.0f}%") #Multiplies by 100 and hides decimal points to print a clean, whole percentage (e.g., 75%)
print(f"  Fresh Air Damper: {fresh_damper * 100:.0f}%")
print(f"  Return Temp     : {return_temp}°C")
print(f"  Fresh Air Temp  : {fresh_air_temp}°C")
print("-" * 40)
print(f"  Mixed Air Temp  : {mixed_air_temp:.2f}°C")
print("=" * 40)
```
